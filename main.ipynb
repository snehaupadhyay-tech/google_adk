#### Functions with type hint and docstring

#### Decorators @staticmethod / @classmethod

```python
from datetime import datetime
print("The current time is",datetime.now())
```

```python
def logger(func):
    def wrapper (*args):
        print("The current time is",datetime.now())
        return func(*args)
    return wrapper
```

```python
@logger

def addition(*args):
    return (sum(args))

addition (1,2,3,4,5)
    
```

#### Pydantic basemode

```python
from pydantic import BaseModel

class Employee(BaseModel):
    name : str
    age : int
    salary : float

e1 = Employee(
    name = "abc",
    age = "12",
    salary = "16000"
)
print(e1)
print(type(e1.name))
print(type(e1.age))
print(type(e1.salary))
```
