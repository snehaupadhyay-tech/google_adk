```python
#if sales price is <20, tag it cheap, between 20 and 50 tag as moderate and >50 tag as expensive

def function1(i):
    if i < 20: 
        return 'cheap'
    elif i<50:
        return 'moderate'
    elif i>=50:
        return 'expensive'

df['Affordability'] = df['Sales_Price'].apply(function1)
df

```
