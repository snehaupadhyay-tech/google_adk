```python
df=pd.read_csv('Bookstore.csv')
df
```

```python
df1 = df[['Genre','Sales_Price']]
df1 = df1.groupby('Genre').sum()
df1['Performance'] = df1['Sales_Price'].apply(lambda x:'Yes' if x>50000 else 'No')
df1
```

```python
plt.bar(df1.index,df1['Sales_Price'])
```

```python
sns.barplot(x=df1.index,y=df1['Sales_Price'],hue=df1['Performance'])
```
